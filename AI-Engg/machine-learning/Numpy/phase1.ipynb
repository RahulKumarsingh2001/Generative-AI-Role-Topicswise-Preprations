### Creating Array form List

```python
arr_1d = np.array([1, 2, 3, 4, 5])
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print("1D Array:", arr_1d)
print("2D Array:\n", arr_2d)
```

### List vs Numpy Array

```python
py_list = [1, 2, 3] # list multiplication
print("python list multiplication: ", py_list*2)

np_array = np.array([1, 2, 3]) # element wise multiplication
print("numpy array multiplication: ", np_array*2)
```

### Compare which one is faster Numpy or List

```python
import time

start = time.time()
py_list = [i*2 for i in range(1000000)]
print("List Operation Time: ", time.time() - start)

start = time.time()
np_array = np.arange(1000000) * 2
print("Numpy Operation Time: ", time.time() - start)
```

### Creating Array from scratch

```python
zeros = np.zeros((3,4))
print("Zeros Array:\n", zeros)

ones = np.ones((2,3))
print("\nOnes Array:\n", ones)

full = np.full((2,2), 7) # Print constant value 7 in a 2x2 array
print("\nFull Array:\n", full)

random = np.random.random((2,2)) # Print random values in a 2x2 array
print("\nRandom Array:\n", random)

sequence = np.arange(0, 10, 2) # Print sequence from 0 to 10 with step size of 2
print("\nSequence Array:\n", sequence)
```

### Vector, Matrix and Tensor

```python
vector = np.array([1, 2, 3])
print("Vector:\n", vector)

matrix = np.array([[1, 2, 3], [4, 5, 6]])
print("Matrix:\n", matrix)

tensor = np.array([[[1, 2], [3, 4]],
                   [[5, 6], [7, 8]]])
print("Tensor:\n", tensor)
```

### Array Properties

```python
arr = np.array([[1,2,3], 
                [4,5, "True"]])

print("Shape: ", arr.shape)
print("Size: ", arr.size)
print("Dimensions: ", arr.ndim)
print("Data Type: ", arr.dtype)
```

### Array Reshaping

```python
arr = np.arange(12)
print("Original Array: ", arr)

reshaped = arr.reshape(3, 4)
print("\n Reshaped Array (3x4):\n", reshaped)

flattened = reshaped.flatten()
print("\n Flattened Array: ", flattened)

raveled = reshaped.ravel() # ravel (return view instead of copy) --> return original array
print("\n Raveled Array: ", raveled)

transposed = reshaped.T
print("\n Transposed Array:\n", transposed)
```
