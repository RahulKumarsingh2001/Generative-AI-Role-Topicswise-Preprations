### Apply function

apply() function run a custom function against every value in a columns

```python
# how have more then 5 years of experience we give in a description "Experienced Candidate"

def categorize(num):
    if num > 5:
        return "Experienced candidate"
    elif num <= 5:
        return "Learner candidate"
```

#### groupby() 

it work in three stages --> 

(1) split the data into a group (2) apply the calculation to each group separately (3) then combine the resut back into one table

```python
df["city"].unique()
```

```python
df.groupby("city")['age'].mean()
```

### Types of Aggregation function
.sum(), .mean(), .median(), .min(), .max(), .count(), .size(), .std()

```python
df.groupby("city")['salary'].max()
```

### Grouping multiple Columns and Aggregration Multiple Columns

```python
df.groupby(["department","city"])['salary'].mean()
```

```python
df.groupby("department")[["salary", "performance_score"]].mean() # multiple columns mean for each department
```

```python
df.sample()
```

```python
# Multiple aggregation with agg()

df.groupby("department").agg({
    "salary": ["mean", "max"],
    "performance_score": "mean",
    "employee_id": "count"
})
```

```python
# Named Aggregation

department_summary = df.groupby("department").agg({
    "salary": ["mean", "max"],
    "performance_score": "mean",
    "employee_id": "count"
})
```

### reseting the index 

department_summary.reset_index()
